```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader


def extract_local_pdf(pdf_path: str, output_txt_file: str = "output.txt"):
    # Convert the input path to a Path object for easier file handling
    pdf_path = Path(pdf_path).expanduser()
    # Convert the output path to a Path object
    output_path = Path(output_txt_file).expanduser()

    # Check whether the PDF file exists before trying to read it
    if not pdf_path.exists():
        raise FileNotFoundError(f"PDF file not found: {pdf_path}")

    # Create the output folder if it does not already exist
    output_path.parent.mkdir(parents=True, exist_ok=True)

    # Load the PDF using LangChain's PDF loader
    loader = PyPDFLoader(str(pdf_path))
    docs = loader.load()

    # Write each page's content into the text file
    with output_path.open("w", encoding="utf-8") as f:
        for page_num, doc in enumerate(docs, start=1):
            f.write(f"--- Page {page_num} ---\n")
            f.write(doc.page_content or "")
            f.write("\n\n")

    # Print a message showing the output file was created
    print(f"Saved text from '{pdf_path}' to '{output_path}'.")


# Example usage with your provided paths
extract_local_pdf(r"C:\Users\LENOVO\OneDrive\Desktop\farmerbook.pdf", r"C:\Users\LENOVO\OneDrive\Desktop\output.txt")
```

```python
import os

from langchain_community.document_loaders import (
    TextLoader,
    PyPDFLoader,
    BSHTMLLoader,
    Docx2txtLoader
)
BASE_DIR = r"C:\Users\LENOVO\OneDrive\Desktop\AI_AGRICULTURAL_ADVISOR"

DATA_DIR = os.path.join(
    BASE_DIR,
    "Data"
)
OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "output"
)
# Create output folder if not exists
os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)
print("Data Folder :", DATA_DIR)
print("Output Folder :", OUTPUT_DIR)

LOADERS = {

    ".txt": TextLoader,

    ".pdf": PyPDFLoader,

    ".html": BSHTMLLoader,

    ".docx": Docx2txtLoader

}


def extract_text(file_path):

    extension = os.path.splitext(file_path)[1].lower()


    # Check supported formats

    if extension not in LOADERS:

        print(
            "Unsupported file:",
            file_path
        )

        return

    print(
        "\nProcessing:",
        os.path.basename(file_path)
    )


    # Load document

    loader = LOADERS[extension](
        file_path
    )


    documents = loader.load()



    # Combine pages

    text = ""

    for doc in documents:

        text += doc.page_content

        text += "\n\n"



    # Output filename

    filename = os.path.basename(file_path)

    name, ext = os.path.splitext(filename)



    output_file = os.path.join(

        OUTPUT_DIR,

        f"{name}_{ext.replace('.','')}-output.txt"

    )



    # Save extracted text

    with open(

        output_file,

        "w",

        encoding="utf-8"

    ) as f:

        f.write(text)



    print(
        "Created:",
        output_file
    )


if not os.path.exists(DATA_DIR):

    raise FileNotFoundError(
        f"Data folder not found: {DATA_DIR}"
    )



for file in os.listdir(DATA_DIR):


    file_path = os.path.join(

        DATA_DIR,

        file

    )


    if os.path.isfile(file_path):

        extract_text(file_path)
```
